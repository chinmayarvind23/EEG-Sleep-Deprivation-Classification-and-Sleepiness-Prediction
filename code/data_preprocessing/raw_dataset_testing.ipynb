### Importing Python Libraries

```python
import os
from pathlib import Path
import numpy as np
import pandas as pd
```

### Setting the Dataset Folder Path

```python
dataset_folder = Path("processed_eeg_dataset/")
print("Dataset folder:", dataset_folder)
```

### Defining File Paths

```python
epoch_df_path = dataset_folder / "eeg_epoch_dataframe.pkl"
metadata_csv_path = dataset_folder / "eeg_metadata.csv"
groups_path = dataset_folder / "groups.npy"
x_eeg_path = dataset_folder / "X_eeg.npy"
x_images_path = dataset_folder / "X_images.npy"
y_labels_path = dataset_folder / "y_labels.npy"
```

### Checking Whether All Files Exist

```python
files_dict = {
    "eeg_epoch_dataframe.pkl": epoch_df_path,
    "eeg_metadata.csv": metadata_csv_path,
    "groups.npy": groups_path,
    "X_eeg.npy": x_eeg_path,
    "X_images.npy": x_images_path,
    "y_labels.npy": y_labels_path
}

for file_name, file_path in files_dict.items():
    print(f"{file_name}: {'Found' if file_path.exists() else 'Missing'}")
```

### Loading `eeg_epoch_dataframe.pkl`

```python
eeg_epoch_df = pd.read_pickle(epoch_df_path)
print("eeg_epoch_dataframe loaded successfully")
```

### Checking Whether All Sample Counts Match

```python
sample_counts = {
    "eeg_epoch_dataframe": len(eeg_epoch_df),
    "eeg_metadata": len(eeg_metadata_df),
    "groups": len(groups),
    "X_eeg": len(X_eeg),
    "X_images": len(X_images),
    "y_labels": len(y_labels)
}

all_same = len(set(sample_counts.values())) == 1

print("Do all files have the same number of samples?", all_same)
print("\nSample counts:")
print(sample_counts)
```
